## Exercise 1: List comprehension vs NumPy

```python
import random
import numpy as np

N = 100_000
```

```python
%timeit [random.random() for _ in range(N)]
```

```python
%timeit np.random.random(N)
```

NumPy should be noticeably faster — it generates the whole array in a single vectorized C call instead of looping in Python 100,000 times.

## Exercise 2: Load CSV, display dataframe, plot a chart

Loading `sample_data.csv` — a small comparison of model accuracy vs training time.

```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline

df = pd.read_csv("sample_data.csv")
df
```

Plotting accuracy vs training time per model.

```python
plt.figure(figsize=(8, 4))
plt.bar(df["model"], df["accuracy"])
plt.title("Accuracy by Model")
plt.ylabel("Accuracy")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```
