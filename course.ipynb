```python
# Exercise 2: Stream c4 and count examples processed in 10 seconds
import time

c4_stream = load_dataset("allenai/c4", "en", split="train", streaming=True)

count = 0
start = time.time()
for example in c4_stream:
    count += 1
    if time.time() - start >= 10:
        break

print(f"Processed {count} examples from c4 in {time.time() - start:.1f} seconds")
```

```python
# Exercise 3: Convert a dataset to Parquet and compare the file size to CSV
from pathlib import Path

sample_ds = load_dataset("stanfordnlp/imdb", split="train").select(range(2000))

csv_path = Path("imdb_sample.csv")
parquet_path = Path("imdb_sample.parquet")

sample_ds.to_csv(str(csv_path))
sample_ds.to_parquet(str(parquet_path))

csv_size = csv_path.stat().st_size
parquet_size = parquet_path.stat().st_size

print(f"CSV:     {csv_size:>10,} bytes")
print(f"Parquet: {parquet_size:>10,} bytes")
print(f"Parquet is {csv_size / parquet_size:.1f}x smaller than CSV")
```

```python
# Exercise 4: Create a 70/15/15 train/val/test split with a fixed seed and verify the sizes
full_ds = load_dataset("stanfordnlp/imdb", split="train")

seed = 42
split1 = full_ds.train_test_split(test_size=0.3, seed=seed)
train_ds = split1["train"]

split2 = split1["test"].train_test_split(test_size=0.5, seed=seed)
val_ds = split2["train"]
test_ds = split2["test"]

total = len(train_ds) + len(val_ds) + len(test_ds)
print(f"Train: {len(train_ds)} ({len(train_ds)/total:.1%})")
print(f"Val:   {len(val_ds)} ({len(val_ds)/total:.1%})")
print(f"Test:  {len(test_ds)} ({len(test_ds)/total:.1%})")

# Verify reproducibility: same seed should produce the same split
split1_again = full_ds.train_test_split(test_size=0.3, seed=seed)
assert split1_again["train"][0] == train_ds[0], "Split is not reproducible!"
print("Split is reproducible with seed=42")
```
